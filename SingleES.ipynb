```python
import numpy as np
import pandas as pd

def single_exp_smoothing(d, extra_periods=4, alpha=0.2):
    d = np.array(d, dtype=float)
    cols = len(d)
    
    f = np.full(cols + extra_periods, np.nan)
    
    f[1] = d[0]
    
    for t in range(2, cols):
        f[t] = alpha * d[t-1] + (1 - alpha) * f[t-1]
        
    d_extended = list(d)
    f[cols] = alpha * d_extended[cols-1] + (1 - alpha) * f[cols-1]
    
    for t in range(cols + 1, cols + extra_periods):
        f[t] = alpha * f[t-1] + (1 - alpha) * f[t-1]
        
    df = pd.DataFrame({
        'Demand': d[:cols], 
        'Forecast': f[:cols], 
        'Error': d[:cols] - f[:cols]
    })
    return df, f

sales_data = [
    362, 385, 432, 341, 
    382, 409, 498, 387, 
    473, 513, 582, 474, 
    544, 582, 681, 557, 
    628, 707, 773, 592, 
    627, 725, 854, 661
]

alpha_value = 0.2
df_result, full_forecast = single_exp_smoothing(sales_data, extra_periods=4, alpha=alpha_value)

df_result.index.name = "Period"
print(df_result.head(10))

errors = df_result["Error"].dropna()
MAE = errors.abs().mean()
RMSE = np.sqrt((errors**2).mean())

print(f"\n--- EVALUATION RESULTS (Alpha = {alpha_value}) ---")
print(f"MAE: {round(MAE, 2)}")
print(f"RMSE: {round(RMSE, 2)}")

print("\n--- FORECAST FOR 4 QUARTERS OF YEAR 95 ---")
for i, val in enumerate(full_forecast[len(sales_data):], start=1):
    print(f"Year 95 - Quarter {i}: {round(val, 2)}")
```
