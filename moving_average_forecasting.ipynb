```python
import numpy as np
import pandas as pd

def moving_avg(d, extra_periods=4, n=3):
    d = np.array(d, dtype=float)
    cols = len(d)
    
    f = np.full(cols + extra_periods, np.nan)
    
    for t in range(n, cols):
        f[t] = np.mean(d[t-n:t])
        
    d_extended = list(d)
    for t in range(cols, cols + extra_periods):
        val = np.mean(d_extended[t-n:t])
        f[t] = val
        d_extended.append(val)
        
    df = pd.DataFrame({
        'Demand': d[:cols], 
        'Forecast': f[:cols], 
        'Error': d[:cols] - f[:cols]
    })
    return df, f

sales_data = [
    362, 385, 432, 341, 
    382, 409, 498, 387, 
    473, 513, 582, 474, 
    544, 582, 681, 557, 
    628, 707, 773, 592, 
    627, 725, 854, 661
]

n_periods = 3
df_result, full_forecast = moving_avg(sales_data, extra_periods=4, n=n_periods)

df_result.index.name = "Period"
print(df_result.head(10))

errors = df_result["Error"].dropna()
MAE = errors.abs().mean()
RMSE = np.sqrt((errors**2).mean())

print(f"\n--- EVALUATION RESULTS (n = {n_periods}) ---")
print(f"MAE: {round(MAE, 2)}")
print(f"RMSE: {round(RMSE, 2)}")

print("\n--- FORECAST FOR 4 QUARTERS OF YEAR 95 ---")
for i, val in enumerate(full_forecast[len(sales_data):], start=1):
    print(f"Year 95 - Quarter {i}: {round(val, 2)}")
```
